Pull the three Motor Vehicle Collisions datasets (Crashes, Vehicles, Persons) from 2017 onward, using `fetch_soda` from `data/download.py` so every pull is cached locally as Parquet and logged in `_provenance`.

Each dataset is fetched in its own cell, so a failed or interrupted run only has to redo the dataset it stopped on — `fetch_soda` serves any dataset already in the cache from local Parquet instead of hitting the API again. The first run pulls several million rows (Persons alone is ~5M), so it might take a while, but a local copy will be cached moving forward.

Note: if you use a different WHERE clause/change the parameters in fetch_soda(), new data will be pulled from the API.

see socrata API documentation here:

- Persons: https://dev.socrata.com/foundry/data.cityofnewyork.us/f55k-p6yu
