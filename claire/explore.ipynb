First, fetch data from Open Data Portal (ODP) using the SODA API.

```python
import os
import pandas as pd
from dotenv import load_dotenv
load_dotenv()
from sodapy import Socrata

# Unauthenticated client only works with public data sets. Note 'None'
# in place of application token, and no username or password:
client = Socrata("data.cityofnewyork.us", None)

# results, returned as JSON from API / converted to Python list of dictionaries by sodapy.
FY2027_tickets= client.get_all("pvqr-7yc4", where="violation_code IN (7, 36)")

# Convert to pandas DataFrame
FY2027_tickets_df = pd.DataFrame.from_records(FY2027_tickets)
FY2027_tickets_df.head(10)
```

this took about a minute to fetch all 38k rows. not terrible, but would have to be repeated for other years
I used documentation here: https://dev.socrata.com/foundry/data.cityofnewyork.us/pvqr-7yc4
