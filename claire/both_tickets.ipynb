```python
# combine each fiscal year in DuckDB (UNION ALL BY NAME handles the extra
# column some years have) instead of pd.concat, so ~20M rows never have to
# sit in pandas all at once
from data.base import get_connection

FISCAL_YEAR_TABLES = [
    "tickets_fy2017",
    "tickets_fy2018",
    "tickets_fy2019",
    "tickets_fy2020",
    "tickets_fy2021",
    "tickets_fy2022",
    "tickets_fy2023",
    "tickets_fy2024",
    "tickets_fy2025",
    "tickets_fy2026",
    "tickets_fy2027",
]

con = get_connection()
try:
    union_sql = " UNION ALL BY NAME ".join(
        f"SELECT * FROM {table}" for table in FISCAL_YEAR_TABLES
    )
    con.execute(f"CREATE OR REPLACE TABLE tickets_all AS {union_sql}")
    row_count = con.execute("SELECT COUNT(*) FROM tickets_all").fetchone()[0]
finally:
    con.close()

row_count
```

Running these local pulls to DuckDB took a couple hours, despite efforts to optimize the indexing using a cursor, instead of page scanning. That being said, moving forward, DuckDB is able to execute SQL-like queries 

```python
#do some basic eda of the combined tickets dataset
import pandas as pd

by_year = get_connection().sql("""
    SELECT COUNT(*) AS tickets,
           DATE_PART('year', CAST(issue_date AS TIMESTAMP)) AS year,
           violation_code
    FROM tickets_all
    GROUP BY DATE_PART('year', CAST(issue_date AS TIMESTAMP)), violation_code
    ORDER BY DATE_PART('year', CAST(issue_date AS TIMESTAMP)), violation_code
""").df()

by_year.head(20)
```

```python
#for readability, note that violation_code 7 is "Failure to Stop at Red Light" and violation_code 36 is "School Zone Speed Violation"
violation_code_labels = {
    7: "Failure to Stop at Red Light",
    36: "School Zone Speed Violation"
}
violation_code_colors = {
    7: "#2a78d6",
    36: "#eb6834",
}

#make a grouped bar chart of the number of tickets by year and violation code
import matplotlib.pyplot as plt
import numpy as np

pivot = by_year.pivot(index="year", columns="violation_code", values="tickets")
pivot.columns = pivot.columns.astype(int)

x = np.arange(len(pivot.index))
width = 0.4
codes = list(pivot.columns)

fig, ax = plt.subplots()
for i, code in enumerate(codes):
    offset = (i - (len(codes) - 1) / 2) * width
    ax.bar(x + offset, pivot[code], width, label=violation_code_labels[code], color=violation_code_colors[code])

ax.set_xticks(x, pivot.index.astype(int))
ax.set_xlabel("Year")
ax.set_ylabel("Number of Tickets")
ax.set_title("Number of Tickets by Year and Violation Code")
ax.legend()
plt.show()
```

School zone speeding tickets VASTLY outnumber red light running tickets. This is an extremely valuable piece of context.

```python
# make a bar chart of red light ticket counts by year

red_light_by_year = get_connection().sql("""
    SELECT DATE_PART('year', CAST(issue_date AS TIMESTAMP)) AS year, COUNT(*) AS tickets
    FROM tickets_all
    WHERE CAST(violation_code AS INTEGER) = 7
    GROUP BY 1
    ORDER BY 1
""").df()

import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.bar(red_light_by_year['year'].astype(int).astype(str), red_light_by_year['tickets'], color=violation_code_colors[7])
ax.set_xlabel('Year')
ax.set_ylabel('Number of Red Light Tickets')
ax.set_title('Red Light Tickets by Year')
plt.show()
```

Some ticket entries have a street name, and even fewer also have a cross-street. This could be an easier way of assigning tickets to cameras/intersections than using the street code fields. Let's check if the cross-street is a field that is usually reported for RLC running.

```python
has_street_names = get_connection().sql("""
    SELECT CAST(violation_code AS INTEGER) AS violation_code, COUNT(*) AS tickets
    FROM tickets_all
    WHERE street_name IS NOT NULL AND intersecting_street IS NOT NULL
    GROUP BY violation_code
""").df()

total_tickets = get_connection().sql("""
    SELECT CAST(violation_code AS INTEGER) AS violation_code, COUNT(*) AS tickets
    FROM tickets_all
    GROUP BY violation_code
""").df()

red_light_total = total_tickets.loc[total_tickets['violation_code'] == 7, 'tickets'].sum()
school_zone_total = total_tickets.loc[total_tickets['violation_code'] == 36, 'tickets'].sum()

print("percentage of red light tickets with street names:", (has_street_names[has_street_names['violation_code'] == 7]['tickets'].sum() / red_light_total) * 100)
print("percentage of school zone speed violation tickets with street names:", (has_street_names[has_street_names['violation_code'] == 36]['tickets'].sum() / school_zone_total) * 100)
```

Most camera-issued tickets DO have cross street information. All tickets have at lease a primary street. How does this compare to availability of street code information?

```python
has_street_codes = get_connection().sql("""
    SELECT COUNT(*) AS tickets,
    FROM tickets_all
    WHERE street_code1 <> 0 OR street_code2 <> 0 OR street_code3 <> 0
""").fetchone()[0]

print("Percentage of tickets with street codes:", (has_street_codes / (red_light_total + school_zone_total)) * 100)
```

Overall, way less tickets have street code info to start with!

```python
no_street_names = get_connection().sql("""
    SELECT violation_code, street_code1, street_code2, street_code3, DATE_PART('year', CAST(issue_date AS TIMESTAMP)) AS year
    FROM tickets_all
    WHERE street_name IS NULL OR intersecting_street IS NULL
""").df()

no_street_names.shape
```

```python
street_code_cols = ["street_code1", "street_code2", "street_code3"]
no_street_names["violation_code"] = no_street_names["violation_code"].astype(int)
non_zero_codes = (no_street_names[street_code_cols] != 0).sum(axis=1)
recoverable = no_street_names[non_zero_codes >= 2]

print(
    f"{len(recoverable)} of {len(no_street_names)} tickets missing street name/cross street "
    f"({len(recoverable) / len(no_street_names) * 100:.2f}%) still have at least 2 of 3 street "
    "codes populated, so the location may be recoverable from street codes alone."
)

recoverable.groupby("violation_code").size().rename("recoverable_tickets")
```
