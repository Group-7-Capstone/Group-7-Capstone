```python
# combine each fiscal year as a view, so no second copy is stored (UNION ALL BY NAME handles the extra
# column some years have) instead of pd.concat, so the rows never have to
# sit in pandas all at once
from data.base import get_connection

FISCAL_YEAR_TABLES = [f"rlc_tickets_fy{year}" for year in range(2017, 2028)]

con = get_connection()
try:
    union_sql = " UNION ALL BY NAME ".join(
        f"SELECT * FROM {table}" for table in FISCAL_YEAR_TABLES
    )
    con.execute(f"CREATE OR REPLACE VIEW rlc_tickets_all AS {union_sql}")
    row_count = con.execute("SELECT COUNT(*) FROM rlc_tickets_all").fetchone()[0]
finally:
    con.close()

row_count
```
