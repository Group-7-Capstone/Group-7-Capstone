```python
def get_direction_sample(year, limit=10000):
    start_date = f"{year}-01-01T00:00:00"
    end_date = f"{year + 1}-01-01T00:00:00"

    df = query_nyc_open_data(
        VEHICLE_API,
        {
            "$select": "collision_id,crash_date,travel_direction",
            "$where": (
                f"crash_date >= '{start_date}' "
                f"AND crash_date < '{end_date}'"
            ),
            "$limit": limit
        }
    )

    return df
```

```python
samples = {}

for year in range(2016, 2027):
    print(f"Downloading {year}...")
    samples[year] = get_direction_sample(year)

print("Done!")
```

```python
direction_quality = []

for year, df in samples.items():

    total = len(df)

    missing = df["travel_direction"].isna().sum()

    unknown = (
        df["travel_direction"]
        .astype(str)
        .str.strip()
        .str.lower()
        .eq("unknown")
        .sum()
    )

    direction_quality.append({
        "year": year,
        "sample_size": total,
        "missing_pct": missing / total * 100,
        "unknown_pct": unknown / total * 100
    })

direction_quality = pd.DataFrame(direction_quality)

direction_quality
```

```python
direction_quality = direction_quality[
    direction_quality["year"] >= 2017
].copy()

direction_quality["usable_direction_pct"] = (
    100
    - direction_quality["missing_pct"]
    - direction_quality["unknown_pct"]
)

direction_quality.round(2)
```

```python
linkage_crashes = query_nyc_open_data(
    CRASH_API,
    {
        "$select": """
            collision_id,
            crash_date,
            latitude,
            longitude,
            number_of_persons_injured,
            number_of_persons_killed
        """,
        "$where": (
            "crash_date >= '2025-01-01T00:00:00' "
            "AND crash_date < '2026-01-01T00:00:00'"
        ),
        "$limit": 1000
    }
)

print("Crash rows:", len(linkage_crashes))
print("Unique collision IDs:", linkage_crashes["collision_id"].nunique())
print("Missing collision IDs:", linkage_crashes["collision_id"].isna().sum())
```

```python
linkage_vehicles_all = query_nyc_open_data(
    VEHICLE_API,
    {
        "$select": """
            collision_id,
            vehicle_id,
            travel_direction,
            contributing_factor_1,
            contributing_factor_2,
            pre_crash,
            point_of_impact
        """,
        "$where": (
            "crash_date >= '2025-01-01T00:00:00' "
            "AND crash_date < '2026-01-01T00:00:00'"
        ),
        "$limit": 50000
    }
)

crash_ids = set(linkage_crashes["collision_id"])

linkage_vehicles = linkage_vehicles_all[
    linkage_vehicles_all["collision_id"].isin(crash_ids)
].copy()

print("Vehicle rows matched:", len(linkage_vehicles))
print("Crashes with vehicle records:",
      linkage_vehicles["collision_id"].nunique())

print("Crashes without vehicle records:",
      1000 - linkage_vehicles["collision_id"].nunique())
```

```python
linkage_persons_all = query_nyc_open_data(
    PERSON_API,
    {
        "$select": """
            collision_id,
            person_id,
            vehicle_id,
            person_type,
            person_injury,
            ped_role,
            person_age
        """,
        "$where": (
            "crash_date >= '2025-01-01T00:00:00' "
            "AND crash_date < '2026-01-01T00:00:00'"
        ),
        "$limit": 50000
    }
)

linkage_persons = linkage_persons_all[
    linkage_persons_all["collision_id"].isin(crash_ids)
].copy()

print("Person rows matched:", len(linkage_persons))
print("Crashes with person records:",
      linkage_persons["collision_id"].nunique())

print("Crashes without person records:",
      1000 - linkage_persons["collision_id"].nunique())
```

how many vehicles are involved in one crash:

```python
vehicles_per_crash = (
    linkage_vehicles
    .groupby("collision_id")
    .size()
)

print(vehicles_per_crash.describe())

print("\nVehicle count distribution:")
print(
    vehicles_per_crash
    .value_counts()
    .sort_index()
    .head(10)
)
```

```python
persons_per_crash = (
    linkage_persons
    .groupby("collision_id")
    .size()
)

print(persons_per_crash.describe())

print("\nPerson count distribution:")
print(
    persons_per_crash
    .value_counts()
    .sort_index()
    .head(10)
)
```

```python
vehicle_summary = (
    linkage_vehicles
    .groupby("collision_id")
    .agg(
        n_vehicles=("vehicle_id", "size"),

        any_unsafe_speed=(
            "contributing_factor_1",
            lambda x: x.eq("Unsafe Speed").any()
        ),

        any_traffic_control_disregarded=(
            "contributing_factor_1",
            lambda x: x.eq("Traffic Control Disregarded").any()
        )
    )
    .reset_index()
)

vehicle_summary.head()
```

```python
print("Rows:", len(vehicle_summary))
print(
    "Unique collision IDs:",
    vehicle_summary["collision_id"].nunique()
)

vehicle_summary.head(10)
```

```python
person_summary = (
    linkage_persons
    .groupby("collision_id")
    .agg(
        n_persons=("person_id", "size"),

        n_injured=(
            "person_injury",
            lambda x: x.eq("Injured").sum()
        ),

        n_killed=(
            "person_injury",
            lambda x: x.eq("Killed").sum()
        ),

        n_pedestrians=(
            "person_type",
            lambda x: x.eq("Pedestrian").sum()
        ),

        n_bicyclists=(
            "person_type",
            lambda x: x.eq("Bicyclist").sum()
        )
    )
    .reset_index()
)

person_summary.head()
```

```python
print("Rows:", len(person_summary))
print(
    "Unique collision IDs:",
    person_summary["collision_id"].nunique()
)

person_summary.head(10)
```

```python
crash_master = (
    linkage_crashes
    .merge(
        vehicle_summary,
        on="collision_id",
        how="left"
    )
    .merge(
        person_summary,
        on="collision_id",
        how="left"
    )
)
```

```python
print("Rows:", len(crash_master))
print("Unique collision IDs:", crash_master["collision_id"].nunique())
print("Duplicate collision IDs:", crash_master["collision_id"].duplicated().sum())

crash_master.head(10)
```

```python
vehicle_crash = linkage_vehicles.merge(
    linkage_crashes[
        [
            "collision_id",
            "crash_date",
            "latitude",
            "longitude",
            "number_of_persons_injured",
            "number_of_persons_killed"
        ]
    ],
    on="collision_id",
    how="left",
    suffixes=("_vehicle", "_crash")
)

print("Rows:", len(vehicle_crash))
print(
    "Unique collision IDs:",
    vehicle_crash["collision_id"].nunique()
)
print(
    "Unique vehicle IDs:",
    vehicle_crash["vehicle_id"].nunique()
)

vehicle_crash.head(10)
```

```python
print("Total crashes:", len(crash_master))

print(
    "Missing latitude:",
    crash_master["latitude"].isna().sum()
)

print(
    "Missing longitude:",
    crash_master["longitude"].isna().sum()
)

print(
    "Zero coordinates:",
    (
        (pd.to_numeric(crash_master["latitude"], errors="coerce") == 0) |
        (pd.to_numeric(crash_master["longitude"], errors="coerce") == 0)
    ).sum()
)
```

```python
crash_master["latitude_num"] = pd.to_numeric(
    crash_master["latitude"],
    errors="coerce"
)

crash_master["longitude_num"] = pd.to_numeric(
    crash_master["longitude"],
    errors="coerce"
)

crash_master[
    ["latitude_num", "longitude_num"]
].describe()
```

```python
crash_master["valid_coordinates"] = (
    crash_master["latitude_num"].notna()
    & crash_master["longitude_num"].notna()
    & (crash_master["latitude_num"] != 0)
    & (crash_master["longitude_num"] != 0)
)

crash_master["valid_coordinates"].value_counts()
```

```python
crash_master["valid_coordinates"].value_counts(
    normalize=True
).mul(100).round(2)
```

```python
crash_master["crash_injured"] = pd.to_numeric(
    crash_master["number_of_persons_injured"],
    errors="coerce"
)

crash_master["crash_killed"] = pd.to_numeric(
    crash_master["number_of_persons_killed"],
    errors="coerce"
)
```

```python
consistency_check = crash_master[
    crash_master["n_persons"].notna()
].copy()

consistency_check["injury_match"] = (
    consistency_check["crash_injured"]
    == consistency_check["n_injured"]
)

consistency_check["killed_match"] = (
    consistency_check["crash_killed"]
    == consistency_check["n_killed"]
)
```

```python
print("Crashes compared:", len(consistency_check))

print(
    "Injury matches:",
    consistency_check["injury_match"].sum()
)

print(
    "Injury mismatches:",
    (~consistency_check["injury_match"]).sum()
)

print(
    "Killed matches:",
    consistency_check["killed_match"].sum()
)

print(
    "Killed mismatches:",
    (~consistency_check["killed_match"]).sum()
)
```

```python
vehicle_summary_new = (
    linkage_vehicles
    .assign(
        unsafe_speed=lambda df:
            df["contributing_factor_1"].eq("Unsafe Speed")
            | df["contributing_factor_2"].eq("Unsafe Speed"),

        traffic_control_disregarded=lambda df:
            df["contributing_factor_1"].eq("Traffic Control Disregarded")
            | df["contributing_factor_2"].eq("Traffic Control Disregarded")
    )
    .groupby("collision_id")
    .agg(
        n_vehicles=("vehicle_id", "size"),
        any_unsafe_speed=("unsafe_speed", "any"),
        any_traffic_control_disregarded=(
            "traffic_control_disregarded",
            "any"
        )
    )
    .reset_index()
)
```

```python
print("Rows:", len(vehicle_summary_new))
print(
    "Unique collision IDs:",
    vehicle_summary_new["collision_id"].nunique()
)
print(
    "Crashes with Unsafe Speed:",
    vehicle_summary_new["any_unsafe_speed"].sum()
)
print(
    "Crashes with Traffic Control Disregarded:",
    vehicle_summary_new["any_traffic_control_disregarded"].sum()
)
```

```python
crash_master_new = (
    linkage_crashes
    .merge(
        vehicle_summary_new,
        on="collision_id",
        how="left"
    )
    .merge(
        person_summary,
        on="collision_id",
        how="left"
    )
)
```

```python
crash_master_new["latitude_num"] = pd.to_numeric(
    crash_master_new["latitude"],
    errors="coerce"
)

crash_master_new["longitude_num"] = pd.to_numeric(
    crash_master_new["longitude"],
    errors="coerce"
)

crash_master_new["valid_coordinates"] = (
    crash_master_new["latitude_num"].notna()
    & crash_master_new["longitude_num"].notna()
    & (crash_master_new["latitude_num"] != 0)
    & (crash_master_new["longitude_num"] != 0)
)
```

```python
print("Rows:", len(crash_master_new))
print(
    "Unique collision IDs:",
    crash_master_new["collision_id"].nunique()
)
print(
    "Duplicate collision IDs:",
    crash_master_new["collision_id"].duplicated().sum()
)
print(
    "Valid coordinates:",
    crash_master_new["valid_coordinates"].sum()
)
print(
    "Unsafe Speed crashes:",
    crash_master_new["any_unsafe_speed"].sum()
)
print(
    "Traffic Control Disregarded crashes:",
    crash_master_new["any_traffic_control_disregarded"].sum()
)
```
